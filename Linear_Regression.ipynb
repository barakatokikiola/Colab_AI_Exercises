#2. Select Your Features and Target

They’re good features because they provide information that is directly related to where a property is located, the characteristics of the housing, and the people living there—all of which can influence house prices.

* median_income: higher income often means higher house prices.
* latitude & longitude: capture location, which strongly affects property value.
* housing_median_age: indicates the age/condition of the housing.
* total_rooms & total_bedrooms describe the size and composition of the housing.
* population & households: describe the area's size and density.

Together, they give the model more useful information for predicting median_house_value


```python
from sklearn.preprocessing import StandardScaler
from sklearn.model_selection import train_test_split
from sklearn.linear_model import LinearRegression
from sklearn.metrics import mean_absolute_error, mean_squared_error, r2_score
```

```python

X = df[
    [
        'median_income',
        'latitude',
        'longitude',
        'housing_median_age',
        'total_rooms',
        'total_bedrooms',
        'population',
        'households'
    ]
]

y = df['median_house_value']
```

#3. Split the Dataset

```python
X_train, X_test, y_train, y_test = train_test_split(X, y, test_size=0.3, random_state=42)

# Scale features using StandardScaler
scaler = StandardScaler()
X_train_scaled = scaler.fit_transform(X_train) # Fit training data
X_test_scaled = scaler.transform(X_test) # Transform test using train's stats
```

#4. Train a Linear Regression Model

```python
# Train the model
model = LinearRegression()
model.fit(X_train_scaled, y_train)

# Generate predictions on the test set
y_pred = model.predict(X_test_scaled)
```

#5. Make Predictions and Evaluate

```python
mae = mean_absolute_error(y_test, y_pred)
mse = mean_squared_error(y_test, y_pred)
r2 = r2_score(y_test, y_pred)

print(f"MAE: {mae:,.2f}")
print(f"MSE: {mse:,.2f}")
print(f"R²: {r2:.4f}")
```
