## Using K-means to cluster colors


### Image segmentation

```python
import numpy as np
import matplotlib.pyplot as plt
from sklearn.cluster import KMeans
from skimage import data, io

print("--- Initializing Image Color Segmentation ---")
image =  io.imread("/content/Barakat.jpg")
# Load the built-in image
# image = data.astronaut()
w, h, d = original_shape = tuple(image.shape)

# Flatten the image into a list of pixels
pixels = np.reshape(image, (w * h, d))

# Run KMeans to find five representative color clusters.
n_colors = 5
model = KMeans(n_clusters=n_colors, random_state=42)
print(f"Compressing Image to {n_colors} colors...")
model.fit(pixels)
labels = model.predict(pixels)
compressed_palette = model.cluster_centers_

# Fast way to recreate the image without slow 'for' loops
# We convert it to uint8 so matplotlib can display it correctly
compressed_image = compressed_palette[labels].reshape(w, h, d).astype(np.uint8)

# Display the results side by side
plt.figure(figsize=(10, 5))

# Left side: Original Image
plt.subplot(1, 2, 1)
plt.title('Original Image')
plt.imshow(image)
plt.axis("off")  # Fixed typo here

# Right side: Compressed Image
plt.subplot(1, 2, 2)
plt.title(f'Compressed Image ({n_colors} Colors)')
plt.imshow(compressed_image)
plt.axis("off")  # Fixed typo here

plt.show()
```
